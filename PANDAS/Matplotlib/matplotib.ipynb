```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Matplotlib is a library in Python that is used for data visualization. It provides a wide range of tools and functions to create various types of plots and charts, such as line plots, bar plots, scatter plots, histograms, and more. Matplotlib is highly customizable, allowing users to adjust the appearance of their plots, including colors, labels, titles, and axes. It is widely used in data analysis, scientific research, and machine learning to visualize data and gain insights from it.

x = np.linspace(0, 5, 11)
x
y = x ** 3
y
```

```python
plt.plot(x, y)   # esse graph draw kr sakte hai graph ke under 
plt.xlabel('X Label')  # x axis ka label
plt.ylabel('Y Label')  # y axis ka label
plt.title('Title')  # graph ka title(naam likh sakte hai esse )
```

```python
#subplots :- MAtlab ki ek hi graph ke ander multiple graph draw kr sakte hai




plt.subplot(2, 2, 2)  # 2 row and 2 column me graph draw hoga aur 1st graph draw hoga
plt.plot(x, y)  #esko likhne pr graph ke under graph draw hoga
plt.subplot(2, 2, 1)
plt.plot(y, x)
plt.subplot(2, 2, 3)
plt.plot(x**2 , y**2)  #esse straight line draw hoga graph ke under
plt.subplot(2, 2, 4)
```

```python
#Object Oriented Method :- Isme hum graph ke ander graph draw karte hai aur graph ke ander graph draw karne ke liye hum figure ka use karte hai

fig = plt.Figure(figsize=(20, 8), dpi=100) 

fig = plt.figure()  # esse graph ke ander graph draw hoga
axis1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])  # esse graph ke ander graph draw hoga
axis1.plot(x, y)  # esse graph ke ander graph draw hoga
axis1 = fig.add_axes([0.1, 0.1, 0.4, 0.4])  # esse graph ke ander graph draw hoga
axis1.plot(x, y)  # esse graph ke ander graph draw hoga
```

```python
#Type of Plots :- Matplotlib me alag alag type ke plots draw kr sakte hai jaise ki line plot, bar plot, scatter plot, histogram, pie chart, etc.
plt.scatter(x, y)  # esse scatter plot draw hoga
```

```python
from random import sample
data = sample(range(1, 1000), 100)  # esse random data generate hoga

plt.hist(data)  # esse histogram draw hoga
```
