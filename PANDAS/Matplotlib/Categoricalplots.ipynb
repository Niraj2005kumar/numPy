```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.countplot(x = df['sex'],hue = df['smoker']) #esse hum counting kr sakte hai 
```

```python
sns.barplot(x= df['sex'],y = df['total_bill'],hue = df['smoker']) #isse hum average nikal sakte hai
```

```python
sns.boxplot(x = df['sex'],y = df['total_bill'],hue = df['smoker']) #isse hum average nikal sakte hai
```

```python
sns.violinplot(data=df, x="sex", y="total_bill", hue="smoker", inner="quart", cut=0)
plt.show()
```

```python
sns.stripplot(x = 'day',y = 'total_bill',data=df)
```

```python
sns.swarmplot(x= 'day',y = 'total_bill',data=df)
```

```python
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```
