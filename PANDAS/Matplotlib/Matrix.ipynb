```python
tips = sns.load_dataset('tips')
tips
```

```python
tips = sns.load_dataset("tips")
tipscorr = tips[["total_bill", "tip", "size"]].corr()
tipscorr
tipscorr 
```

```python
sns.heatmap(tipscorr, annot=True, cmap="coolwarm")
```

Cluster map


```python
sns.clustermap(tipscorr, annot=True, cmap="coolwarm")
```

Pivot table heat maps


```python
pvtflight = flights.pivot_table(values='passengers', index='month', 
columns='year')
pvtflight
```

```python
sns.heatmap(pvtflight)
```
