```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Name": ["Niraj", "Rahul", "Aman", "Priya"],
    "Department": ["CSE", "ECE", "ME", "CE"]
})

marks = pd.DataFrame({
    "Student_ID": [1, 2, 3, 5],
    "Marks": [85, 78, 92, 88]
    

})

print(students)
print(marks)
```

```python
pd.merge(students, marks, on='Student_ID',how = 'outer')
```

```python
pd.merge(students, marks, on='Student_ID',how = 'left')
```

```python
#Concatination of dataframes

import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Niraj", "Rahul", "Aman"],
    "Marks": [85, 78, 92]
})

df2 = pd.DataFrame({
    "Name": ["Priya", "Neha", "Riya"],
    "Marks": [88, 90, 95]
})



```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```

```python
df1 = pd.DataFrame({
    "Name": ["Niraj", "Rahul", "Aman"],
    "Marks": [85, 78, 92]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "City": ["Ranchi", "Patna", "Delhi"],
    "Age": [21, 22, 20]
}, index=[1, 2, 3])
```

```python
df1
```

```python
df1.join(df2)
```
