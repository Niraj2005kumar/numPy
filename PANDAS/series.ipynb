```python
#Selecting Row

import pandas as pd

data = {
    "Roll_No": [101, 102, 103, 104],
    "Name": ["Rahul", "Priya", "Aman", "Neha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.loc[[0,1,2]]
```

```python
df.iloc[3]
```

```python
df.loc[[2,3]][["Age","Name"]]
```

```python
# conditional Selection

# i only want to see those people whose age is above 30

df[df["Age"]>20]
```

```python
# i only want people whose age is abhove 30 and their cirt must be pari

df[(df["Age"] > 21) & (df["Marks"]=='21')]
```
