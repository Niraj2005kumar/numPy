```python
import numpy as np
import pandas as pd
```

```python

data = {
    "Name": ["Rahul", "Priya", "Aman", "Neha"],
    "Age": [20, 21, np.nan, 22],
    "Marks": [85, np.nan, 90, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.isna()  
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
# Remove Missing Data :- esko remove krne ke liye rows ke basis pr depend krta hai


df
```

```python
df.dropna() #dropna krne se value ko remove kr sakte hia esko use krte hi dara data hi remove ho jayega 
```

```python
df.dropna(thresh=1)  #thresh ka matlab yeh hai ki mere data me 3 null value rhna chahiye 
```

```python
# Filling missing data  : - eska matlab yah hia ki missing value ko fill kr sakte hai 

df.fillna(0)


```

```python
values = {'Age':45,'Marks':100}
df.fillna(value=values)
```

```python
#agar hum kise ek particuller value ke under ki value ko fill krne chahiye toh 

df.fillna(df.mean())
```
